```python
import json
from transformers import pipeline
```

```python
# Initialize BART summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
```

```python
# Example
ARTICLE  = "Olivia: Who are you voting for in this election? \r\nOliver: Liberals as always.\r\nOlivia: Me too!!\r\nOliver: Great"
print(summarizer(ARTICLE, max_length=50, min_length=10, do_sample=False))
```

```python
# Load JSON file: path to original summary
with open("original_summary.json","r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
# Collect lengths of human-generated summaries
summary_lengths = [len(entry.get("summary", "").split()) for entry in data if entry.get("summary")]

# Calculate min and max lengths, then extend by 10%
min_length = int(min(summary_lengths) * 0.9)
max_length = int(max(summary_lengths) * 1.1)

print(f"Suggested min_length: {min_length}, max_length: {max_length}")
```

```python
# Process each dialogue to generate summaries
output_data = []
for entry in data:
    dialogue = entry.get("dialogue", "")
    id = entry.get("id", "")
    if dialogue:
        summary = summarizer(dialogue,
                             max_length=63,
                             min_length = 2,
                             do_sample = False)[0]["summary_text"]
        output_data.append({
            "id": id,
            "dialogue": dialogue,
            "summary": summary
        })
```

```python
# Save the output data to a new JSON file
with open("summarized_dialogues.json", "w", encoding="utf-8") as f:
    json.dump(output_data, f, ensure_ascii=False, indent=4)

print("Summaries generated and saved to summarized_dialogues.json.")
```
