```python
import json
import openai
```

```python
# Your OpenAI API key
openai.api_key = "YOUR_API_KEYS"
```

```python
# Load JSON file
with open("original_summary.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
# Use gpt-3.5-turbo to generate summaries
output_data = []
for entry in data:
    dialogue = entry.get("dialogue", "")
    id_ = entry.get("id", "")
    if dialogue:
        # Create prompt
        prompt = f"Summarize the following conversation:\n{dialogue}\nPlease return only the summarized text."
        try:
            response = openai.ChatCompletion.create(
                model="gpt-4-turbo",
                messages=[
                    {"role": "system", "content": "You are a helpful assistant that summarizes conversations."},
                    {"role": "user", "content": prompt}
                ],
                max_tokens=63
            )

            summary = response.choices[0].message.content.strip()
            output_data.append({
                "id": id_,
                "dialogue": dialogue,
                "summary": summary
            })
        except Exception as e:
            print(f"Error generating summary for id {id_}: {e}")
```

```python
# Save summaries to file
with open("results.json", "w", encoding="utf-8") as f:
    json.dump(output_data, f, ensure_ascii=False, indent=4)

print("Summaries generated using GPT-3.5-turbo and saved to results.json.")
```
